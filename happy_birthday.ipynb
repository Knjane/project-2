```python
!pip install -q ColabTurtlePlus

import ColabTurtlePlus.Turtle as t
import random

# Initialize ColabTurtlePlus screen and settings
t.clearscreen()
t.setup(500, 500)
t.bgcolor("#08091c")
t.speed(30)
t.hideturtle()

for i in range(45):
    t.penup()
    t.goto(0, -100)
    t.setheading(random.randint(30, 150))
    t.pencolor(random.choice(["gold", "cyan", "orange", "tomato", "lime"]))
    t.pendown()
    t.forward(random.randint(20, 180))
    t.dot(7, random.choice(["gold", "cyan", "orange", "tomato", "lime"]))

t.penup()
t.goto(0, 150)
t.color("gold")
t.write("HAPPY BIRTHDAY", align="center", font=("Arial", 30, "bold"))
t.goto(0, -200)
t.color("tomato")
t.write("wishing you a blast", align="center", font=("Arial", 16, "bold"))
```
